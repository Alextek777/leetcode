[1266. Minimum Time Visiting All Points](https://leetcode.com/problems/minimum-time-visiting-all-points/description/?envType=daily-question&envId=2026-01-12)

```python
from typing import List

class Solution:
    def minTimeToVisitAllPoints(self, points: List[List[int]]) -> int:
        result = 0

        for i in range(0, len(points) - 1):
            dx = abs(points[i + 1][0] - points[i][0])
            dy = abs(points[i + 1][1] - points[i][1])
            result += max(dx, dy) # in other words result += min(dx, dy)  - (max(dx, dy) - min(dx, dy)) we travel min() on diagonal and what is left travel rather horizontally or vertically
        return result
```

```python
s = Solution()

s.minTimeToVisitAllPoints([[1,1],[3,4],[-1,0]])
```
